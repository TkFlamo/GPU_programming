```python
from PIL import Image
import sys
import os

def convert_jpg_to_bmp(input_path: str, output_path: str) -> bool:
    try:
        print(f"Converting {input_path} to BMP...")
        
        with Image.open(input_path) as img:
            if img.mode != 'RGB':
                img = img.convert('RGB')
            
            img.save(output_path, 'BMP')
        
        input_size = os.path.getsize(input_path)
        output_size = os.path.getsize(output_path)
        
        print(f"  Successfully converted to {output_path}")
        print(f"  Input size: {input_size:,} bytes")
        print(f"  Output size: {output_size:,} bytes")
        print(f"  Size ratio: {output_size/input_size:.2f}x")
        
        return True
        
    except FileNotFoundError:
        print(f"Error: File not found - {input_path}")
        return False
    except Exception as e:
        print(f"Error converting {input_path}: {str(e)}")
        return False

def convert_bmp_to_jpg(input_path: str, output_path: str, quality: int = 95) -> bool:
    try:
        print(f"Converting {input_path} to JPG...")
        
        with Image.open(input_path) as img:
            if img.mode != 'RGB':
                img = img.convert('RGB')
            
            img.save(output_path, 'JPEG', quality=quality, optimize=True)
        
        input_size = os.path.getsize(input_path)
        output_size = os.path.getsize(output_path)
        
        print(f"  Successfully converted to {output_path}")
        print(f"  Input size: {input_size:,} bytes")
        print(f"  Output size: {output_size:,} bytes")
        print(f"  Size ratio: {output_size/input_size:.2f}x")
        print(f"  JPG quality: {quality}")
        
        return True
        
    except FileNotFoundError:
        print(f"Error: File not found - {input_path}")
        return False
    except Exception as e:
        print(f"Error converting {input_path}: {str(e)}")
        return False
```
